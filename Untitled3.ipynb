```python
sum_even = 0
count = 0

while count < 10:
    num = int(input())

    if num == 0:
        break

    if num < 0:
        continue

    if num > 0 and num % 2 == 0:
        sum_even += num

    count += 1

print(sum_even)
```

```python
num = int(input())
for i in range(1, 13):
    product = num * i

    if product % 3 == 0:
        continue

    if product > 50:
        break

    print(product)
```

```python
secret = 7
attempts = 0

while True:
    guess = int(input())

    if guess < 1 or guess > 20:
        continue

    attempts += 1

    if guess == secret:
        break

print(attempts)
```

```python
n = int(input())
count = 0

for i in range(1, n + 1):

    if i > 50:
        break

    if i % 8 == 0:
        continue

    if i % 4 == 0:
        count += 1

print(count)
```

```python
for i in range(1, 6):
    for j in range(1, i + 1):

        if j == 3:
            continue

        if i * j > 6:
            break

        print(j, end=" ")

    print()
```
